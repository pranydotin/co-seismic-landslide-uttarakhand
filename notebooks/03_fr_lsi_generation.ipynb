```python
import os
import glob
import numpy as np
import rasterio
```

```python
input_folder = "/content/drive/MyDrive/cosesmic/layers/output_weight_rasters_aligned"
output_path = "/content/drive/MyDrive/cosesmic/layers/LSI_map.tif"
```

```python
files = sorted(glob.glob(os.path.join(input_folder, "W_*.tif")))

print(f"Found {len(files)} weight rasters")
```

```python
with rasterio.open(files[0]) as src0:
    profile = src0.profile.copy()
    nodata = src0.nodata if src0.nodata is not None else -9999

    lsi = np.zeros(src0.shape, dtype=np.float32)
```

```python
for f in files:
    with rasterio.open(f) as src:
        print(os.path.basename(f), src.shape)
```

```python
for f in files:
    print(f"Adding: {os.path.basename(f)}")

    with rasterio.open(f) as src:
        data = src.read(1).astype(np.float32)

        # Replace nodata with 0 (important)
        data[data == src.nodata] = 0

        lsi += data
```

```python
profile.update(dtype=rasterio.float32, nodata=-9999, compress="LZW")

with rasterio.open(output_path, "w", **profile) as dst:
    dst.write(lsi, 1)

print(f"\nLSI map saved at: {output_path}")
```
