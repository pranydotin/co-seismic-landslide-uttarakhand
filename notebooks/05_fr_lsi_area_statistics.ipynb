```python
raster_path = "/content/drive/MyDrive/cosesmic/layers/LSI_classes2.tif"
```

```python
with rasterio.open(raster_path) as src:
    data = src.read(1)
    nodata = src.nodata
    pixel_size_x = src.res[0]
    pixel_size_y = src.res[1]
```

```python

data = data[data != nodata]
```

```python
classes, counts = np.unique(data, return_counts=True)
```

```python
pixel_area_m2 = pixel_size_x * pixel_size_y

# km²
areas_km2 = (counts * pixel_area_m2) / 1e6
```

```python

total_area = areas_km2.sum()

percentages = (areas_km2 / total_area) * 100
```

```python
class_labels = {
    1: "Very Low",
    2: "Low",
    3: "Moderate",
    4: "High",
    5: "Very High"
}

labels = [class_labels.get(int(c), str(c)) for c in classes]
```

```python
df = pd.DataFrame({
    "Class_Value": classes,
    "Susceptibility": labels,
    "Pixel_Count": counts,
    "Area_km2": areas_km2,
    "Percentage": percentages
})
```

```python
df["Area_km2"] = df["Area_km2"].round(2)
df["Percentage"] = df["Percentage"].round(2)
```
